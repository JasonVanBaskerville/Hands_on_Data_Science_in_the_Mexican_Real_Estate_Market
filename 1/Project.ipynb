3. Pembersihan dataset 1 (df1)

- .dropna(): menghapus baris yang memiliki nilai kosong (sisa 583 baris).
- .assign(price_usd = lambda x: ...): membuat ulang kolom harga.
- .str.replace("$", "") dan .str.replace(",", ""): menghapus simbol dolar dan koma dari harga.
- .astype(float): mengubah harga dari teks menjadi angka desimal.
- df1.shape[0], df1["price_usd"].dtype, df1.isnull().sum().sum(): memeriksa hasil (583 baris, tipe float64, tidak ada nilai kosong).

```python
# Step 1: Drop missing values
(
    pd.read_csv(path_1)
    .dropna()  # Remove rows where values are missing
    
)
```

```python
# Step 2: The Complete Chain
(
    pd.read_csv(path_1)
    # 1. Drop rows with missing critical data
    .dropna()
    
    # 2. Create clean numeric columns using `.assign()`
    # We use 'lambda x' to refer to the data at this specific step in the chain
    # We overwrite 'price_usd' by accessing the current df (x in our lambda function) and cleaning the string
    .assign(
        price_usd = lambda x: x["price_usd"]
                          .str.replace("$", "", regex=False)
                          .str.replace(",", "", regex=False)
                          .astype(float)
    )
)
```

```python
df1 = (
    pd.read_csv(path_1)
    # 1. Drop rows with missing critical data
    .dropna()
    
    # 2. Create clean numeric columns using `.assign()`
    # We use 'lambda x' to refer to the data at this specific step in the chain
    # We overwrite 'price_usd' by accessing the current df (x in our lambda function) and cleaning the string
    .assign(
        price_usd = lambda x: x["price_usd"]
                          .str.replace("$", "", regex=False)
                          .str.replace(",", "", regex=False)
                          .astype(float)
    )
)
```

5. Pembersihan dataset 3 (df3)

- Pada dataset 3, kolom lat-lon berisi koordinat dalam satu teks, dan place_with_parent_names berisi nama wilayah yang dipisah tanda |.
- .str.split(",", expand=True): memisahkan koordinat menjadi dua kolom. [0] menjadi lat, [1] menjadi lon, keduanya diubah ke float.
- .str.split("|", expand=True)[2]: mengambil bagian ke-3 sebagai nama negara bagian (state).
- .drop(columns=[...]): menghapus dua kolom asal yang sudah diurai.
- .dropna(): menghapus baris kosong (hasil 582 baris).

```python
(
    pd.read_csv(path_3)
    .head()
)
```

```python
(
    pd.read_csv(path_3)
    ["place_with_parent_names"]
    .str.split("|", expand=True)
    .head()
)
```

```python
# cleaning process using method chaining
df3 = (
    pd.read_csv(path_3)
    # Drop null values from df3
    .dropna()
    # Create "lat" and "lon" columns for df3
    .assign(lat=lambda x: x["lat-lon"]
                .str.split(",", expand=True)[0]           
                .astype(float),
            lon=lambda x: x["lat-lon"]
                .str.split(",", expand=True)[1]           
                .astype(float),
            # Create "state" column for df3
            state=lambda x: x["place_with_parent_names"]
            .str.split("|", expand=True)[2]
           )
    
    # Drop "place_with_parent_names" and "lat-lon" from df3
    .drop(columns=["place_with_parent_names", "lat-lon"])
            
    
)

# Print object type, shape, and head
print("df3 type:", type(df3))
print("df3 shape:", df3.shape)
df3.head()
```

6. Menggabungkan dataset

- set(df1) == set(df2) == set(df3): memeriksa bahwa ketiga DataFrame punya kolom yang sama. Hasilnya cocok.
- pd.concat([df1, df2, df3], ignore_index=True): menumpuk ketiga data secara vertikal menjadi satu df (1.736 baris, 6 kolom).
- df.shape, df.isnull().sum(), df.index: memverifikasi ukuran, tidak ada nilai kosong, dan indeks berurutan.

```python
# Check if all column sets are identical
columns_match = set(df1) == set(df2) == set(df3)

if columns_match:
    print("✅ All DataFrames have the same columns.")
else:
    print("❌ Column mismatch detected!")
    # Optional: See the difference
    print("df1 vs df2 diff:", set(df1.columns) ^ set(df2.columns))
    print("df2 vs df3 diff:", set(df2.columns) ^ set(df3.columns))
```

```python
# Stack df1, df2, and df3 vertically
df = pd.concat([df1, df2, df3], ignore_index=True)

# Verify the result
print(f"Combined Shape: {df.shape}")
df.head()
```
